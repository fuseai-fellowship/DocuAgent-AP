```python
from paddleocr import LayoutDetection 
layout_model = LayoutDetection(
    model_name="PP-DocLayout_plus-L",
    threshold={2: 0.2, 12: 0.2, 14: 0.2},  # text, header, footer
    device='gpu:0',
    layout_nms=True,
    layout_merge_bboxes_mode= "large",
    layout_unclip_ratio=1.1
)
```

```python
import cv2 

def preprocess_region(img, x1, y1, x2, y2, pad_w=5, pad_h=5, min_height=32):
      # Add padding with boundary checks
    x1_pad = max(0, x1 - pad_w)
    y1_pad = max(0, y1 - pad_h)
    x2_pad = min(img.shape[1], x2 + pad_w)
    y2_pad = min(img.shape[0], y2 + pad_h)
    
    # Crop region
    region = img[y1_pad:y2_pad, x1_pad:x2_pad]
    
    # Upscale if region is too small
    h, w = region.shape[:2]
    if h < min_height:
        scale = min_height / h
        new_width = int(w * scale)
        region = cv2.resize(region, (new_width, min_height), interpolation=cv2.INTER_LINEAR)
   
    return region
```

```python
import cv2
import os
import time

def test_crop_based_pipeline(path, layout_model, table_pipeline=None, output_dir="output/crop_based"):
    now = time.time()

    
    # --- Load image ---
    img = cv2.imread(path)
    print(img.shape)
    if img is None:
        raise FileNotFoundError(f"Could not load image at path: {path}")

    # --- Prepare output folders ---
    crop_dir = os.path.join(output_dir, "crop")
    os.makedirs(crop_dir, exist_ok=True)

    # --- Run layout model ---
    layout_results = layout_model.predict(img)
    crop_index = 0

    for result in layout_results:
        for box in result["boxes"]:
            cls = box["label"]
            coords = [int(x) for x in box["coordinate"]]
            x1, y1, x2, y2 = coords

            # # --- Chart / Image recognition ---
            # if cls in ["chart", "image"]:
            #     region = preprocess_region(img, x1, y1, x2, y2)
            #     crop_path = os.path.join(crop_dir, f"{cls}_region_{crop_index}.png")
            #     cv2.imwrite(crop_path, region)

            #     # Run specialized model
            #     results = model.predict(
            #     input={"image": region}, 
            #     batch_size=1
            #         )
                
            #     for res in results:
            #         res.print()
            #         res.save_to_json(crop_dir)  # Save structured outputs
               
            #     crop_index += 1

              #  --- Table recognition ---
            if cls == "table":
                region = preprocess_region(img, x1, y1, x2, y2)
                output = table_pipeline.predict(
                    input=region,
                    use_doc_orientation_classify=False,
                    use_doc_unwarping=False,
                )

                for res in output:
                    res.print()
                    res.save_to_img(crop_dir)
                    res.save_to_xlsx(crop_dir)
                    res.save_to_html(crop_dir)
                    res.save_to_json(crop_dir)

    print(f"Total runtime: {time.time() - now:.2f}s")
```
